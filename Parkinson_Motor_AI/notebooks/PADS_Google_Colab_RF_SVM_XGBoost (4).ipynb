```python

from pathlib import Path
from zipfile import ZipFile
import shutil

starter = Path("/content/PADS_RF_SVM_XGBoost_Starter.zip")
project = Path("/content/PADS_RF_SVM_XGBoost_Starter")

assert starter.exists(), "Starter ZIP missing!"

project.mkdir(parents=True, exist_ok=True)

prefix = "PADS_RF_SVM_XGBoost_Starter/"
allowed_folders = ("pads_motor/", "scripts/", "tests/")
allowed_files = {"requirements.txt", "README.md"}

restored = []

with ZipFile(starter) as z:
    assert z.testzip() is None, "ZIP corrupted"

    for member in z.infolist():
        name = member.filename

        if not name.startswith(prefix) or member.is_dir():
            continue

        relative = name[len(prefix):]

        if not (
            relative.startswith(allowed_folders)
            or relative in allowed_files
        ):
            continue

        destination = (project / relative).resolve()

        if not destination.is_relative_to(project.resolve()):
            raise ValueError("Unsafe ZIP path")

        destination.parent.mkdir(
            parents=True, exist_ok=True
        )

        if not destination.exists():
            with z.open(member) as source:
                with open(destination, "wb") as target:
                    shutil.copyfileobj(source, target)

        restored.append(relative)

print("PREPROCESSING SOURCE CODE RESTORED!")
print("\nRestored source files:")

for filename in restored:
    print(" -", filename)

print("\nFINAL VERIFICATION")
print(
    "Preprocessing script:",
    (project / "scripts/prepare.py").exists()
)
print(
    "Feature extraction script:",
    (project / "pads_motor/features.py").exists()
)
print(
    "Hybrid CNN model:",
    (project / "outputs/hybrid_cnn/PADS_Hybrid_CNN.pt").exists()
)
```

```python

from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED

raw = Path(
    "/content/PADS_RF_SVM_XGBoost_Starter/data/pads_dataset"
)

export = Path("/content/PADS_Raw_Export_Parts")
export.mkdir(parents=True, exist_ok=True)

assert (raw / "patients").exists()
assert (raw / "movement").exists()

files = sorted(
    p for p in raw.rglob("*") if p.is_file()
)

# Maximum uncompressed data per ZIP part: 200 MB
MAX_BYTES = 200 * 1024 * 1024

parts = []
current_files = []
current_bytes = 0

for file in files:
    size = file.stat().st_size

    if current_files and current_bytes + size > MAX_BYTES:
        parts.append(current_files)
        current_files = []
        current_bytes = 0

    current_files.append(file)
    current_bytes += size

if current_files:
    parts.append(current_files)

prefix = Path(
    "Parkinson_Motor_AI/data/raw/pads_dataset"
)

print("RAW PADS EXPORT STARTED")
print("Total raw files:", len(files))
print("ZIP parts:", len(parts))

for i, group in enumerate(parts, start=1):

    zip_path = export / f"PADS_Raw_Part_{i:02d}.zip"

    with ZipFile(
        zip_path, "w", ZIP_DEFLATED, compresslevel=3
    ) as z:

        for file in group:
            archive_path = prefix / file.relative_to(raw)

            z.write(
                file,
                archive_path.as_posix()
            )

    print(
        zip_path.name,
        "-",
        round(zip_path.stat().st_size / 1e6, 2),
        "MB",
        "-",
        len(group),
        "files",
        flush=True
    )

print("\nRAW DATASET BACKUP PARTS CREATED!")
print("Saved folder:", export)
```

```python

import numpy as np
import pandas as pd
import torch
from torch import nn
from pathlib import Path

ROOT = Path("/content/Parkinson_Motor_AI")

# 1. Load saved Hybrid CNN checkpoint
checkpoint = torch.load(
    ROOT / "models/Hybrid_CNN/PADS_Hybrid_CNN.pt",
    map_location="cpu",
    weights_only=False
)

# 2. Rebuild original Hybrid architecture
class HybridMotorCNN(nn.Module):
    def __init__(self, n_features=18):
        super().__init__()

        self.cnn = nn.Sequential(
            nn.Conv1d(6, 32, 9, padding=4),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.MaxPool1d(4),
            nn.Conv1d(32, 64, 7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(4),
            nn.Conv1d(64, 64, 5, padding=2),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1),
            nn.Flatten()
        )

        self.handcrafted = nn.Sequential(
            nn.Linear(n_features, 32),
            nn.ReLU()
        )

        self.fusion = nn.Sequential(
            nn.Linear(96, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 1)
        )

    def forward(self, sensor, manual):
        deep = self.cnn(sensor)
        hand = self.handcrafted(manual)
        return self.fusion(torch.cat([deep, hand], dim=1)).squeeze(1)

model = HybridMotorCNN(
    len(checkpoint["feature_names"])
)

model.load_state_dict(checkpoint["model_state"], strict=True)
model.eval()

print("HYBRID CNN MODEL LOADED!")

# 3. Load the processed data
data = ROOT / "data/processed"

signals = np.load(
    data / "signals.npy",
    mmap_mode="r"
)

features = pd.read_csv(
    data / "features.csv",
    dtype={"participant_id": str}
)

metadata = pd.read_csv(
    data / "metadata.csv",
    dtype={"participant_id": str}
)

splits = pd.read_csv(
    data / "participant_splits.csv",
    dtype={"participant_id": str}
)

assert len(signals) == len(features) == len(metadata)
assert signals.shape[1:] == (900, 6)
assert (
    features["participant_id"].tolist()
    == metadata["participant_id"].tolist()
)

# Choose one previously validated participant
split_map = splits.set_index("participant_id")["split"]
metadata["split"] = metadata["participant_id"].map(split_map)

val_ids = metadata.loc[
    metadata["split"] == "val", "participant_id"
].unique()

assert len(val_ids) == 54

participant = sorted(val_ids)[0]

indices = np.flatnonzero(
    metadata["participant_id"].eq(participant).to_numpy()
)

# 4. Apply saved training-only normalization
raw = np.asarray(signals[indices], dtype=np.float32)

mean = np.asarray(
    checkpoint["signal_mean"], dtype=np.float32
).reshape(1, 1, 6)

std = np.asarray(
    checkpoint["signal_std"], dtype=np.float32
).reshape(1, 1, 6)

sensor = ((raw - mean) / np.maximum(std, 1e-6))
sensor = sensor.transpose(0, 2, 1).copy()

cols = checkpoint["feature_names"]

manual = features.iloc[indices][cols].to_numpy(
    dtype=np.float32
)

manual = (
    manual - checkpoint["feature_mean"]
) / np.maximum(checkpoint["feature_std"], 1e-6)

# 5. Run inference on CPU
with torch.no_grad():
    logits = model(
        torch.from_numpy(sensor),
        torch.from_numpy(manual.astype(np.float32))
    )

    recording_probabilities = (
        torch.sigmoid(logits).numpy()
    )

# Match our previous participant-level evaluation
prediction = float(recording_probabilities.mean())

print("\nHYBRID CNN INFERENCE TEST")
print("Validation participant:", participant)
print("Recordings:", len(indices))
print("Research prediction score:", round(prediction, 4))
print("Predicted class:", "PD" if prediction >= 0.50 else "Healthy")

# 6. Verify against previously saved validation prediction
saved = pd.read_csv(
    ROOT / "results/hybrid_cnn/hybrid_validation_subjects.csv",
    dtype={"participant_id": str}
)

reference = saved.loc[
    saved["participant_id"] == participant,
    "probability_pd"
]

assert len(reference) == 1

difference = abs(prediction - float(reference.iloc[0]))

print("Saved-prediction difference:", round(difference, 8))

assert difference < 0.005, "Prediction verification failed"

print("\nHYBRID CNN CLOUD INFERENCE VERIFIED!")
print("Research classification only — not clinical PD probability")
```

```python

import copy
import json
import numpy as np
import pandas as pd
import torch
from pathlib import Path
from torch import nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.metrics import (
    balanced_accuracy_score, roc_auc_score,
    recall_score, f1_score
)

torch.manual_seed(42)
np.random.seed(42)
torch.set_num_threads(2)

root = Path("/content/PADS_RF_SVM_XGBoost_Starter")
out = root / "outputs"
cnn_dir = out / "cnn"
cnn_dir.mkdir(exist_ok=True)

# Load original PADS signals and official grouped splits
raw = np.load(out / "signals.npy")
meta = pd.read_csv(
    out / "metadata.csv",
    dtype={"participant_id": str}
)
splits = pd.read_csv(
    out / "participant_splits.csv",
    dtype={"participant_id": str}
)

assert len(raw) == len(meta)
assert raw.shape[1:] == (900, 6)

split_map = splits.set_index("participant_id")["split"]
meta["split"] = meta["participant_id"].map(split_map)
assert meta["split"].notna().all()

tr = np.where(meta["split"].eq("train"))[0]
va = np.where(meta["split"].eq("val"))[0]

assert set(meta.iloc[tr].participant_id).isdisjoint(
    set(meta.iloc[va].participant_id)
)

y_train = meta.iloc[tr]["label"].to_numpy(np.float32)
y_val = meta.iloc[va]["label"].to_numpy(np.float32)

# Normalize using TRAINING recordings only
mu = raw[tr].mean(axis=(0, 1))
sd = raw[tr].std(axis=(0, 1)).clip(1e-5)

def prepare(indices):
    x = ((raw[indices] - mu) / sd).astype(np.float32)
    return torch.from_numpy(x.transpose(0, 2, 1).copy())

X_train = prepare(tr)
X_val = prepare(va)

train_loader = DataLoader(
    TensorDataset(X_train, torch.from_numpy(y_train)),
    batch_size=64,
    shuffle=True
)

val_loader = DataLoader(
    TensorDataset(X_val, torch.from_numpy(y_val)),
    batch_size=128
)

# Compact CNN suitable for a first experiment
class MotorCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv1d(6, 32, kernel_size=9, padding=4),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.MaxPool1d(4),

            nn.Conv1d(32, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(4),

            nn.Conv1d(64, 64, kernel_size=5, padding=2),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(64, 1)
        )

    def forward(self, x):
        return self.classifier(self.features(x)).squeeze(1)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
model = MotorCNN().to(device)

# Account for training class imbalance
n0 = int((y_train == 0).sum())
n1 = int((y_train == 1).sum())

loss_fn = nn.BCEWithLogitsLoss(
    pos_weight=torch.tensor(
        [n0 / n1], device=device
    )
)
optimizer = torch.optim.Adam(
    model.parameters(), lr=0.001, weight_decay=1e-4
)

print("CNN TRAINING STARTED")
print("Device:", device)
print("Training recordings:", len(tr))
print("Validation recordings:", len(va))
print("Input shape:", tuple(X_train.shape))

best_loss = float("inf")
best_state = None
history = []
patience_count = 0

for epoch in range(1, 16):
    model.train()
    train_losses = []

    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        logits = model(xb)
        loss = loss_fn(logits, yb)
        loss.backward()
        optimizer.step()
        train_losses.append(loss.item())

    model.eval()
    val_losses = []
    with torch.no_grad():
        for xb, yb in val_loader:
            xb, yb = xb.to(device), yb.to(device)
            val_losses.append(
                loss_fn(model(xb), yb).item()
            )

    vl = float(np.mean(val_losses))
    tl = float(np.mean(train_losses))
    history.append({
        "epoch": epoch,
        "train_loss": tl,
        "val_loss": vl
    })

    print(
        f"Epoch {epoch:02d} | "
        f"train loss={tl:.4f} | "
        f"val loss={vl:.4f}"
    )

    if vl < best_loss - 0.0001:
        best_loss = vl
        best_state = copy.deepcopy(model.state_dict())
        patience_count = 0
    else:
        patience_count += 1

    if patience_count >= 4:
        print("Early stopping")
        break

# Evaluate on VALIDATION participants only
model.load_state_dict(best_state)
model.eval()

probabilities = []

with torch.no_grad():
    for xb, _ in val_loader:
        logits = model(xb.to(device))
        probabilities.extend(
            torch.sigmoid(logits).cpu().numpy().tolist()
        )

val_df = meta.iloc[va][
    ["participant_id", "label"]
].copy()
val_df["probability_pd"] = probabilities

subject_df = (
    val_df.groupby(
        ["participant_id", "label"],
        as_index=False
    )["probability_pd"].mean()
)

actual = subject_df["label"].to_numpy()
probs = subject_df["probability_pd"].to_numpy()
predicted = (probs >= 0.5).astype(int)

print("\n1D-CNN VALIDATION RESULTS")
print("Participants:", len(subject_df))
print(
    "Balanced Accuracy:",
    round(balanced_accuracy_score(actual, predicted), 3)
)
print(
    "ROC-AUC:",
    round(roc_auc_score(actual, probs), 3)
)
print(
    "PD Sensitivity:",
    round(recall_score(actual, predicted, pos_label=1), 3)
)
print(
    "Healthy Specificity:",
    round(recall_score(actual, predicted, pos_label=0), 3)
)
print(
    "Macro F1:",
    round(f1_score(actual, predicted, average="macro"), 3)
)

# Save model, normalization and evaluation files
torch.save({
    "model_state": model.cpu().state_dict(),
    "normalization_mean": mu,
    "normalization_std": sd,
    "channels": ["ax", "ay", "az", "gx", "gy", "gz"],
    "sampling_rate_hz": 100,
    "window_samples": 900
}, cnn_dir / "PADS_1D_CNN.pt")

subject_df.to_csv(
    cnn_dir / "cnn_validation_subjects.csv",
    index=False
)
pd.DataFrame(history).to_csv(
    cnn_dir / "cnn_training_history.csv",
    index=False
)

print("\nCNN experiment completed!")
print("Files saved in:", cnn_dir)
```

```python

import copy
import numpy as np
import pandas as pd
import torch

from pathlib import Path
from torch import nn
from torch.utils.data import TensorDataset, DataLoader
from sklearn.metrics import (
    balanced_accuracy_score,
    roc_auc_score,
    recall_score,
    f1_score
)

torch.manual_seed(42)
np.random.seed(42)
torch.set_num_threads(2)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# 1. Load the same PADS dataset
root = Path(
    "/content/PADS_RF_SVM_XGBoost_Starter/outputs"
)

required = [
    "signals.npy",
    "metadata.csv",
    "features.csv",
    "participant_splits.csv"
]

for filename in required:
    assert (root / filename).exists(), (
        f"Missing: {filename}. Run PADS Step 1-6 first."
    )

raw = np.load(root / "signals.npy", mmap_mode="r")

meta = pd.read_csv(
    root / "metadata.csv",
    dtype={"participant_id": str}
)

features = pd.read_csv(
    root / "features.csv",
    dtype={"participant_id": str}
)

splits = pd.read_csv(
    root / "participant_splits.csv",
    dtype={"participant_id": str}
)

# Check raw signals and handcrafted features align
assert len(raw) == len(meta) == len(features)
assert raw.shape[1:] == (900, 6)

for column in [
    "participant_id", "label",
    "source", "task", "wrist"
]:
    assert (
        meta[column].astype(str).tolist()
        == features[column].astype(str).tolist()
    ), f"Record mismatch: {column}"

assert splits["participant_id"].is_unique

meta["split"] = meta["participant_id"].map(
    splits.set_index("participant_id")["split"]
)

assert meta["split"].notna().all()

train_idx = np.flatnonzero(
    meta["split"].eq("train").to_numpy()
)
val_idx = np.flatnonzero(
    meta["split"].eq("val").to_numpy()
)

assert set(
    meta.iloc[train_idx]["participant_id"]
).isdisjoint(
    set(meta.iloc[val_idx]["participant_id"])
)

# 2. Select handcrafted sensor features only
# Excludes diagnosis, participant IDs and metadata
feature_cols = [
    c for c in features.columns
    if c.startswith(("acc_", "gyro_", "jerk_"))
]

assert len(feature_cols) == 18, feature_cols

F = features[feature_cols].to_numpy(
    dtype=np.float32
)

assert np.isfinite(F).all()

# 3. Normalize using TRAINING data only
raw_train = np.asarray(
    raw[train_idx], dtype=np.float32
)

signal_mean = raw_train.mean(axis=(0, 1))
signal_std = np.maximum(
    raw_train.std(axis=(0, 1)), 1e-6
)

feature_mean = F[train_idx].mean(axis=0)
feature_std = np.maximum(
    F[train_idx].std(axis=0), 1e-6
)

def create_dataset(indices):

    sensor = np.asarray(
        raw[indices], dtype=np.float32
    )

    sensor = (
        (sensor - signal_mean) / signal_std
    ).transpose(0, 2, 1).copy()

    engineered = (
        (F[indices] - feature_mean) / feature_std
    ).copy()

    labels = meta.iloc[indices][
        "label"
    ].to_numpy(dtype=np.float32)

    return TensorDataset(
        torch.from_numpy(sensor),
        torch.from_numpy(engineered),
        torch.from_numpy(labels)
    )

train_data = create_dataset(train_idx)
val_data = create_dataset(val_idx)

train_loader = DataLoader(
    train_data, batch_size=64, shuffle=True
)

val_loader = DataLoader(
    val_data, batch_size=128, shuffle=False
)

# 4. Hybrid CNN architecture
class HybridMotorCNN(nn.Module):

    def __init__(self, n_features):
        super().__init__()

        # Branch A: learn temporal sensor features
        self.cnn = nn.Sequential(
            nn.Conv1d(6, 32, 9, padding=4),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.MaxPool1d(4),

            nn.Conv1d(32, 64, 7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(4),

            nn.Conv1d(64, 64, 5, padding=2),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1),
            nn.Flatten()
        )

        # Branch B: learn from handcrafted features
        self.handcrafted = nn.Sequential(
            nn.Linear(n_features, 32),
            nn.ReLU()
        )

        # Fuse the two branches
        self.fusion = nn.Sequential(
            nn.Linear(96, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 1)
        )

    def forward(self, sensor, engineered):

        deep = self.cnn(sensor)
        manual = self.handcrafted(engineered)

        combined = torch.cat(
            [deep, manual], dim=1
        )

        return self.fusion(combined).squeeze(1)

model = HybridMotorCNN(
    len(feature_cols)
).to(device)

# 5. Train using the same binary labels
y_train = meta.iloc[train_idx]["label"].to_numpy()

healthy = int((y_train == 0).sum())
pd_count = int((y_train == 1).sum())

loss_fn = nn.BCEWithLogitsLoss(
    pos_weight=torch.tensor(
        healthy / pd_count,
        dtype=torch.float32,
        device=device
    )
)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

print("HYBRID CNN TRAINING STARTED")
print("Device:", device)
print("Training recordings:", len(train_idx))
print("Validation recordings:", len(val_idx))
print("Handcrafted features:", len(feature_cols))

best_loss = float("inf")
best_state = None
patience = 0
history = []

for epoch in range(1, 16):

    model.train()
    total_train_loss = 0.0

    for sensor, engineered, labels in train_loader:

        sensor = sensor.to(device)
        engineered = engineered.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        logits = model(sensor, engineered)
        loss = loss_fn(logits, labels)

        loss.backward()
        optimizer.step()

        total_train_loss += loss.item() * len(labels)

    model.eval()
    total_val_loss = 0.0

    with torch.no_grad():

        for sensor, engineered, labels in val_loader:

            sensor = sensor.to(device)
            engineered = engineered.to(device)
            labels = labels.to(device)

            logits = model(sensor, engineered)
            loss = loss_fn(logits, labels)

            total_val_loss += loss.item() * len(labels)

    train_loss = total_train_loss / len(train_data)
    val_loss = total_val_loss / len(val_data)

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss
    })

    print(
        f"Epoch {epoch:02d} | "
        f"train={train_loss:.4f} | "
        f"val={val_loss:.4f}",
        flush=True
    )

    if val_loss < best_loss - 0.0001:

        best_loss = val_loss
        best_state = copy.deepcopy(
            model.state_dict()
        )
        patience = 0

    else:
        patience += 1

        if patience >= 4:
            print("Early stopping")
            break

# 6. Participant-level validation
model.load_state_dict(best_state)
model.eval()

probabilities = []

with torch.no_grad():

    for sensor, engineered, _ in val_loader:

        logits = model(
            sensor.to(device),
            engineered.to(device)
        )

        probabilities.extend(
            torch.sigmoid(logits)
            .cpu().numpy().tolist()
        )

val_rows = meta.iloc[val_idx][
    ["participant_id", "label"]
].copy()

val_rows["probability_pd"] = probabilities

subject_results = (
    val_rows.groupby(
        ["participant_id", "label"],
        as_index=False
    )["probability_pd"].mean()
)

actual = subject_results["label"].to_numpy()
probs = subject_results[
    "probability_pd"
].to_numpy()

predicted = (probs >= 0.5).astype(int)

print("\nHYBRID CNN VALIDATION RESULTS")
print("Participants:", len(subject_results))

print(
    "Balanced Accuracy:",
    round(balanced_accuracy_score(
        actual, predicted
    ), 3)
)

print(
    "ROC-AUC:",
    round(roc_auc_score(actual, probs), 3)
)

print(
    "PD Sensitivity:",
    round(recall_score(
        actual, predicted, pos_label=1
    ), 3)
)

print(
    "Healthy Specificity:",
    round(recall_score(
        actual, predicted, pos_label=0
    ), 3)
)

print(
    "Macro F1:",
    round(f1_score(
        actual, predicted, average="macro"
    ), 3)
)

# 7. Save model and preprocessing information
save_dir = root / "hybrid_cnn"
save_dir.mkdir(parents=True, exist_ok=True)

torch.save({
    "model_state": model.cpu().state_dict(),
    "feature_names": feature_cols,
    "signal_mean": signal_mean,
    "signal_std": signal_std,
    "feature_mean": feature_mean,
    "feature_std": feature_std,
    "sampling_rate_hz": 100,
    "raw_window_samples": 900,
    "model_type": "HybridMotorCNN"
}, save_dir / "PADS_Hybrid_CNN.pt")

subject_results.to_csv(
    save_dir / "hybrid_validation_subjects.csv",
    index=False
)

pd.DataFrame(history).to_csv(
    save_dir / "hybrid_training_history.csv",
    index=False
)

print("\nHYBRID EXPERIMENT COMPLETED!")
print("Model saved in:", save_dir)
```

```python

import numpy as np
from scipy.signal import find_peaks

# Fixed device test durations
TS_SECONDS = 10
BS_SECONDS = 10
RS_SECONDS = 10
GI_SECONDS = 180

GRAVITY = 9.81

# DEMONSTRATION THRESHOLDS ONLY
# NOT clinically validated or device calibrated
DEMO_THRESHOLDS = {
    "TS": [0.12, 0.24, 0.40, 0.70],
    "BS": [210, 165, 120, 75],
    "RS": [2.4, 1.8, 1.2, 0.6],
    "GI": [110, 95, 80, 65]
}

def higher_is_worse(value, thresholds):
    return sum(value >= t for t in thresholds)

def lower_is_worse(value, thresholds):
    return sum(value < t for t in thresholds)

def extract_motor_features(
    acc_xyz_ms2,
    gyro_xyz_rad_s,
    tap_times_s,
    step_count,
    sampling_rate=100
):
    acc = np.asarray(acc_xyz_ms2, dtype=float)
    gyro = np.asarray(gyro_xyz_rad_s, dtype=float)
    taps = np.asarray(tap_times_s, dtype=float)

    expected = 10 * sampling_rate

    if acc.shape != (expected, 3):
        raise ValueError("TS requires 10 seconds of XYZ acceleration")

    if gyro.shape != (expected, 3):
        raise ValueError("RS requires 10 seconds of XYZ gyroscope")

    if not np.isfinite(acc).all() or not np.isfinite(gyro).all():
        raise ValueError("Invalid sensor values")

    if np.any(np.diff(taps) <= 0):
        raise ValueError("Tap timestamps must be increasing")

    if np.any(taps < 0) or np.any(taps >= BS_SECONDS):
        raise ValueError("Taps must be within 10 seconds")

    # TS: supervisor's acceleration RMS equation
    acc_magnitude = np.linalg.norm(acc, axis=1)
    dynamic_acc = acc_magnitude - GRAVITY
    tremor_rms = np.sqrt(np.mean(dynamic_acc**2))

    # RS: angular velocity magnitude and mean cycle peak
    omega = np.linalg.norm(gyro, axis=1)

    # Demo cycle detection; needs device validation
    peaks, _ = find_peaks(
        omega,
        distance=int(sampling_rate * 0.45),
        prominence=0.2
    )

    if len(peaks) == 0:
        raise ValueError("No valid RS movement cycles detected")

    mean_gyro_peak = np.mean(omega[peaks])

    # BS: original tap-rate and interval equations
    tap_rate = 60 * len(taps) / BS_SECONDS
    tap_intervals = np.diff(taps)

    # GI: original cadence equation
    # Step count supplied by demo / future step detector
    if not isinstance(step_count, (int, np.integer)) or step_count < 0:
        raise ValueError("Step count must be nonnegative integer")

    cadence = 60 * step_count / GI_SECONDS

    return {
        "tremor_rms_ms2": float(tremor_rms),
        "mean_gyro_peak_rad_s": float(mean_gyro_peak),
        "tap_rate_per_min": float(tap_rate),
        "tap_interval_std_s": (
            float(np.std(tap_intervals))
            if len(tap_intervals) else None
        ),
        "cadence_steps_min": float(cadence)
    }

def generate_demo_scores(features):
    return {
        "TS": higher_is_worse(
            features["tremor_rms_ms2"],
            DEMO_THRESHOLDS["TS"]
        ),
        "BS": lower_is_worse(
            features["tap_rate_per_min"],
            DEMO_THRESHOLDS["BS"]
        ),
        "RS": lower_is_worse(
            features["mean_gyro_peak_rad_s"],
            DEMO_THRESHOLDS["RS"]
        ),
        "GI": lower_is_worse(
            features["cadence_steps_min"],
            DEMO_THRESHOLDS["GI"]
        )
    }

def calculate_mps(scores):
    assert set(scores) == {"TS", "BS", "RS", "GI"}
    assert all(type(v) is int and 0 <= v <= 4
               for v in scores.values())
    return sum(scores.values()) / 16.0

# SIMULATED SENSOR READINGS — NOT REAL PATIENT DATA
fs = 100
t = np.arange(0, 10, 1 / fs)

acc_demo = np.column_stack([
    0.20 * np.sin(2 * np.pi * 5 * t),
    0.10 * np.cos(2 * np.pi * 5 * t),
    GRAVITY + 0.15 * np.sin(2 * np.pi * 5 * t)
])

gyro_demo = np.column_stack([
    1.8 * np.sin(2 * np.pi * 1.2 * t),
    0.10 * np.cos(2 * np.pi * 1.2 * t),
    0.05 * np.sin(2 * np.pi * 1.2 * t)
])

tap_times_demo = np.arange(0.3, 10, 0.48)
step_count_demo = 288

features = extract_motor_features(
    acc_demo,
    gyro_demo,
    tap_times_demo,
    step_count_demo,
    sampling_rate=fs
)

scores = generate_demo_scores(features)
mps = calculate_mps(scores)

print("MOTOR ASSESSMENT - DEMO MODE")
print("NOT CLINICALLY CALIBRATED")
print("-" * 40)

print("\nEXTRACTED FEATURES")
for name, value in features.items():
    print(name, ":", value)

print("\nAUTOMATIC DEMO MOTOR SCORES")
for name, value in scores.items():
    print(f"{name}: {value}/4")

print(f"\nTEMPORARY MPS: {mps:.4f}")
print("MPS range: 0 to 1")
print("Not a Parkinson's diagnosis or PD probability")
```

```python

import numpy as np
from scipy.signal import butter, sosfiltfilt, find_peaks

# GI must use the full 180-second recording
GI_SECONDS = 180
fs = 100

def detect_gait_steps(acc_xyz_ms2, sampling_rate=100):

    acc = np.asarray(acc_xyz_ms2, dtype=float)

    expected_samples = GI_SECONDS * sampling_rate

    if acc.shape != (expected_samples, 3):
        raise ValueError(
            f"GI requires {expected_samples} samples "
            f"({GI_SECONDS} seconds, 3 axes)"
        )

    if not np.isfinite(acc).all():
        raise ValueError("Invalid acceleration values")

    # Acceleration magnitude
    magnitude = np.linalg.norm(acc, axis=1)

    # Demonstration walking-frequency filter
    sos = butter(
        3,
        [0.6, 3.0],
        btype="bandpass",
        fs=sampling_rate,
        output="sos"
    )

    filtered = sosfiltfilt(sos, magnitude)

    # Candidate steps from acceleration peaks
    peaks, _ = find_peaks(
        filtered,
        distance=int(0.4 * sampling_rate),
        prominence=0.3
    )

    step_times = peaks / sampling_rate
    step_count = len(peaks)

    cadence = 60 * step_count / GI_SECONDS

    intervals = np.diff(step_times)

    variability = (
        float(np.std(intervals))
        if len(intervals) > 0 else None
    )

    return {
        "step_count": int(step_count),
        "cadence_steps_min": float(cadence),
        "step_interval_std_s": variability
    }


# SIMULATED 180-SECOND WRIST ACCELEROMETER DATA
# This is not a real patient recording.

t_gi = np.arange(0, GI_SECONDS, 1 / fs)
rng = np.random.default_rng(42)

gi_acc_demo = np.column_stack([
    0.15 * np.sin(2 * np.pi * 1.6 * t_gi),
    0.10 * np.cos(2 * np.pi * 1.6 * t_gi),
    9.81
    + 1.2 * np.sin(2 * np.pi * 1.6 * t_gi)
    + 0.05 * rng.standard_normal(len(t_gi))
])

# Automatically detect candidate steps
gait = detect_gait_steps(
    gi_acc_demo,
    sampling_rate=fs
)

# Update previous assessment with detected gait features
features["step_count"] = gait["step_count"]
features["cadence_steps_min"] = gait["cadence_steps_min"]
features["step_interval_std_s"] = gait["step_interval_std_s"]

# Recalculate GI score automatically using DEMO thresholds
scores["GI"] = lower_is_worse(
    gait["cadence_steps_min"],
    DEMO_THRESHOLDS["GI"]
)

# Recalculate temporary MPS
mps = calculate_mps(scores)

print("AUTOMATIC GI DETECTION - DEMO MODE")
print("----------------------------------")
print("Recording duration:", GI_SECONDS, "seconds")
print("Detected step candidates:", gait["step_count"])
print("Cadence:", round(gait["cadence_steps_min"], 2))
print("Step interval variability:", gait["step_interval_std_s"])

print("\nMOTOR SCORES (DEMO ONLY)")
for name, score in scores.items():
    print(f"{name}: {score}/4")

print("\nTEMPORARY MPS:", round(mps, 4))
print("STATUS: UNCALIBRATED DEMONSTRATION")
```

```python

import numpy as np
import pandas as pd
import joblib
from pathlib import Path

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from xgboost import XGBClassifier

ROOT = Path("/content/PADS_RF_SVM_XGBoost_Starter/outputs")
OUT = ROOT / "final_model_comparison"
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "features.csv",
    "metadata.csv",
    "participant_splits.csv",
    "cnn/cnn_validation_subjects.csv",
    "hybrid_cnn/hybrid_validation_subjects.csv"
]

missing = [f for f in required if not (ROOT / f).exists()]
if missing:
    raise FileNotFoundError(
        "Missing files. Restore PADS outputs first:\n"
        + "\n".join(missing)
    )

feat = pd.read_csv(
    ROOT / "features.csv",
    dtype={"participant_id": str}
)

meta = pd.read_csv(
    ROOT / "metadata.csv",
    dtype={"participant_id": str}
)

splits = pd.read_csv(
    ROOT / "participant_splits.csv",
    dtype={"participant_id": str}
)

assert len(feat) == len(meta)
assert splits["participant_id"].is_unique

for c in ["participant_id", "label"]:
    assert (
        feat[c].astype(str).tolist()
        == meta[c].astype(str).tolist()
    ), f"Alignment mismatch: {c}"

feat["split"] = feat["participant_id"].map(
    splits.set_index("participant_id")["split"]
)

assert feat["split"].notna().all()

# Only the 18 handcrafted sensor features
feature_cols = [
    c for c in feat.columns
    if c.startswith(("acc_", "gyro_", "jerk_"))
]

assert len(feature_cols) == 18, feature_cols

train = feat[feat["split"] == "train"].copy()
val = feat[feat["split"] == "val"].copy()

assert set(train.participant_id).isdisjoint(
    set(val.participant_id)
)

X_train = train[feature_cols].to_numpy(dtype=float)
y_train = train["label"].to_numpy(dtype=int)

X_val = val[feature_cols].to_numpy(dtype=float)

assert np.isfinite(X_train).all()
assert np.isfinite(X_val).all()
assert set(np.unique(y_train)) == {0, 1}

hc = int((y_train == 0).sum())
pd_count = int((y_train == 1).sum())

# Fresh classical baselines
models = {
    "RF": RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=2,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ),

    "SVM": make_pipeline(
        StandardScaler(),
        SVC(
            kernel="rbf",
            C=1.0,
            class_weight="balanced",
            probability=True,
            random_state=42
        )
    ),

    "XGBoost": XGBClassifier(
        n_estimators=250,
        max_depth=3,
        learning_rate=0.05,
        subsample=0.85,
        colsample_bytree=0.9,
        scale_pos_weight=hc / pd_count,
        objective="binary:logistic",
        eval_metric="logloss",
        tree_method="hist",
        random_state=42,
        n_jobs=2
    )
}

print("TRAINING PARTICIPANTS:", train.participant_id.nunique())
print("VALIDATION PARTICIPANTS:", val.participant_id.nunique())
print("FEATURES:", len(feature_cols))

for name, model in models.items():

    print("\nTraining:", name, flush=True)
    model.fit(X_train, y_train)

    # Recording-level probabilities
    prob = model.predict_proba(X_val)[:, 1]

    pred = val[
        ["participant_id", "label"]
    ].copy()

    pred["probability_pd"] = prob

    # SAME participant-level aggregation as CNN
    subject_pred = pred.groupby(
        ["participant_id", "label"],
        as_index=False
    )["probability_pd"].mean()

    subject_pred.to_csv(
        OUT / f"{name}_validation_subjects.csv",
        index=False
    )

    joblib.dump(
        model,
        OUT / f"{name}_model.joblib"
    )

    print(
        name,
        "- saved",
        len(subject_pred),
        "participant predictions"
    )

joblib.dump(feature_cols, OUT / "feature_columns.joblib")

print("\nCLASSICAL VALIDATION PREDICTIONS READY!")
```

```python

from sklearn.metrics import (
    balanced_accuracy_score,
    roc_auc_score,
    recall_score,
    f1_score
)
import json

MODEL_FILES = {
    "RF": OUT / "RF_validation_subjects.csv",
    "SVM": OUT / "SVM_validation_subjects.csv",
    "XGBoost": OUT / "XGBoost_validation_subjects.csv",
    "1D-CNN": ROOT / "cnn/cnn_validation_subjects.csv",
    "Hybrid CNN": ROOT / "hybrid_cnn/hybrid_validation_subjects.csv"
}

expected = (
    val[["participant_id", "label"]]
    .drop_duplicates()
    .sort_values("participant_id")
    .reset_index(drop=True)
)

assert len(expected) == 54

rows = []
locked_thresholds = {}

for name, file_path in MODEL_FILES.items():

    df = pd.read_csv(
        file_path,
        dtype={"participant_id": str}
    ).sort_values("participant_id").reset_index(drop=True)

    assert len(df) == len(expected), name
    assert df.participant_id.tolist() == (
        expected.participant_id.tolist()
    ), f"Different participants: {name}"

    assert df.label.astype(int).tolist() == (
        expected.label.astype(int).tolist()
    ), f"Label mismatch: {name}"

    y = df["label"].to_numpy(dtype=int)
    p = df["probability_pd"].to_numpy(dtype=float)

    assert np.isfinite(p).all()
    assert np.all((p >= 0) & (p <= 1))

    # Untuned, common 0.50 threshold
    pred_fixed = (p >= 0.5).astype(int)

    ba_fixed = balanced_accuracy_score(
        y, pred_fixed
    )

    # Threshold selection on VALIDATION ONLY
    candidates = np.round(
        np.arange(0.10, 0.901, 0.01), 2
    )

    candidate_scores = [
        balanced_accuracy_score(
            y, (p >= t).astype(int)
        )
        for t in candidates
    ]

    # Tie-breaker: threshold closest to 0.5
    best_index = sorted(
        range(len(candidates)),
        key=lambda i: (
            -candidate_scores[i],
            abs(candidates[i] - 0.5)
        )
    )[0]

    threshold = float(candidates[best_index])
    locked_thresholds[name] = threshold

    predicted = (p >= threshold).astype(int)

    rows.append({
        "Model": name,
        "BA_at_0.5": ba_fixed,
        "Val_Threshold": threshold,
        "BA_Val_Tuned": balanced_accuracy_score(
            y, predicted
        ),
        "ROC_AUC": roc_auc_score(y, p),
        "PD_Sensitivity": recall_score(
            y, predicted, pos_label=1
        ),
        "HC_Specificity": recall_score(
            y, predicted, pos_label=0
        ),
        "Macro_F1": f1_score(
            y, predicted, average="macro"
        )
    })

comparison = pd.DataFrame(rows)

comparison = comparison.sort_values(
    ["BA_Val_Tuned", "ROC_AUC"],
    ascending=[False, False]
).reset_index(drop=True)

comparison.to_csv(
    OUT / "five_model_validation_comparison.csv",
    index=False
)

selection = {
    "leading_validation_candidate": comparison.iloc[0]["Model"],
    "thresholds_selected_on": "validation_only",
    "locked_thresholds": locked_thresholds,
    "test_used_for_selection": False,
    "clinical_scoring_model": False
}

with open(OUT / "selection_config.json", "w") as f:
    json.dump(selection, f, indent=2)

print("FIVE-MODEL VALIDATION COMPARISON")
print("Participants:", len(expected))
print()
print(comparison.round(3).to_string(index=False))

print("\nLEADING VALIDATION CANDIDATE:")
print(selection["leading_validation_candidate"])

print("\nThresholds and results saved in:", OUT)
print("TEST SET NOT EVALUATED BY THIS CODE")
```

```python

import numpy as np
import pandas as pd
import joblib
import torch
from torch import nn
from pathlib import Path

ROOT = Path("/content/PADS_RF_SVM_XGBoost_Starter/outputs")
OUT = ROOT / "final_model_comparison"

# Reconstruct the ORIGINAL saved 1D-CNN architecture
class PADS1DCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv1d(6, 32, 9, padding=4),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.MaxPool1d(4),
            nn.Conv1d(32, 64, 7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(4),
            nn.Conv1d(64, 64, 5, padding=2),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(64, 1)
        )

    def forward(self, x):
        return self.classifier(self.features(x)).squeeze(1)


# Reconstruct the ORIGINAL saved Hybrid CNN
class HybridMotorCNN(nn.Module):
    def __init__(self, n_features=18):
        super().__init__()

        self.cnn = nn.Sequential(
            nn.Conv1d(6, 32, 9, padding=4),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.MaxPool1d(4),
            nn.Conv1d(32, 64, 7, padding=3),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(4),
            nn.Conv1d(64, 64, 5, padding=2),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1),
            nn.Flatten()
        )

        self.handcrafted = nn.Sequential(
            nn.Linear(n_features, 32),
            nn.ReLU()
        )

        self.fusion = nn.Sequential(
            nn.Linear(96, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 1)
        )

    def forward(self, x, engineered):
        deep = self.cnn(x)
        manual = self.handcrafted(engineered)
        combined = torch.cat([deep, manual], dim=1)
        return self.fusion(combined).squeeze(1)


# Load checkpoints without training again
cnn_ckpt = torch.load(
    ROOT / "cnn/PADS_1D_CNN.pt",
    map_location="cpu",
    weights_only=False
)

hybrid_ckpt = torch.load(
    ROOT / "hybrid_cnn/PADS_Hybrid_CNN.pt",
    map_location="cpu",
    weights_only=False
)

cnn_model = PADS1DCNN()
cnn_model.load_state_dict(cnn_ckpt["model_state"], strict=True)
cnn_model.eval()

hybrid_model = HybridMotorCNN(
    len(hybrid_ckpt["feature_names"])
)
hybrid_model.load_state_dict(
    hybrid_ckpt["model_state"], strict=True
)
hybrid_model.eval()

classical = {
    name: joblib.load(OUT / f"{name}_model.joblib")
    for name in ["RF", "SVM", "XGBoost"]
}

feature_cols = joblib.load(
    OUT / "feature_columns.joblib"
)

feat = pd.read_csv(
    ROOT / "features.csv",
    dtype={"participant_id": str}
)

meta = pd.read_csv(
    ROOT / "metadata.csv",
    dtype={"participant_id": str}
)

splits = pd.read_csv(
    ROOT / "participant_splits.csv",
    dtype={"participant_id": str}
)

assert len(meta) == len(feat)
assert splits.participant_id.is_unique

for col in ["participant_id", "label"]:
    assert (
        meta[col].astype(str).tolist()
        == feat[col].astype(str).tolist()
    ), f"Data mismatch: {col}"

meta["split"] = meta["participant_id"].map(
    splits.set_index("participant_id")["split"]
)

assert meta["split"].notna().all()

signals = np.load(
    ROOT / "signals.npy",
    mmap_mode="r"
)

assert signals.shape == (len(meta), 900, 6)

def normalize_signals(raw, mean, std):
    mean = np.asarray(
        mean, dtype=np.float32
    ).reshape(1, 1, 6)

    std = np.asarray(
        std, dtype=np.float32
    ).reshape(1, 1, 6)

    x = (raw - mean) / np.maximum(std, 1e-6)
    return x.transpose(0, 2, 1).copy()


def torch_probabilities(model, sensor, engineered=None):
    results = []

    with torch.no_grad():
        for start in range(0, len(sensor), 128):
            end = start + 128

            batch = torch.from_numpy(sensor[start:end])

            if engineered is None:
                logits = model(batch)
            else:
                extra = torch.from_numpy(
                    engineered[start:end]
                )
                logits = model(batch, extra)

            results.extend(
                torch.sigmoid(logits).numpy().tolist()
            )

    return np.asarray(results)


def predict_subjects(model_name, indices):

    if model_name in classical:

        x = feat.iloc[indices][
            feature_cols
        ].to_numpy(dtype=float)

        probabilities = classical[
            model_name
        ].predict_proba(x)[:, 1]

    else:
        raw = np.asarray(
            signals[indices],
            dtype=np.float32
        )

        if model_name == "1D-CNN":
            sensor = normalize_signals(
                raw,
                cnn_ckpt["normalization_mean"],
                cnn_ckpt["normalization_std"]
            )
            probabilities = torch_probabilities(
                cnn_model, sensor
            )

        elif model_name == "Hybrid CNN":
            sensor = normalize_signals(
                raw,
                hybrid_ckpt["signal_mean"],
                hybrid_ckpt["signal_std"]
            )

            manual = feat.iloc[indices][
                hybrid_ckpt["feature_names"]
            ].to_numpy(dtype=np.float32)

            manual = (
                manual - hybrid_ckpt["feature_mean"]
            ) / np.maximum(
                hybrid_ckpt["feature_std"], 1e-6
            )

            probabilities = torch_probabilities(
                hybrid_model,
                sensor,
                manual.astype(np.float32)
            )

        else:
            raise ValueError(model_name)

    rows = meta.iloc[indices][
        ["participant_id", "label"]
    ].copy()

    rows["probability_pd"] = probabilities

    return (
        rows.groupby(
            ["participant_id", "label"],
            as_index=False
        )["probability_pd"]
        .mean()
        .sort_values("participant_id")
        .reset_index(drop=True)
    )


print("ALL FIVE SAVED MODELS LOADED SUCCESSFULLY!")
print("CNN architecture verified")
print("Hybrid architecture verified")
print("No model retraining performed")
```

```python

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.metrics import roc_auc_score

ROOT = Path(
    "/content/PADS_RF_SVM_XGBoost_Starter/outputs"
)
OUT = ROOT / "final_model_comparison"

names = [
    "RF", "SVM", "XGBoost",
    "1D-CNN", "Hybrid CNN"
]

# Load saved locked-test predictions
data = {}

for name in names:

    filename = (
        name.replace(" ", "_") + "_test_subjects.csv"
    )

    df = pd.read_csv(
        OUT / filename,
        dtype={"participant_id": str}
    )

    df = df.sort_values(
        "participant_id"
    ).reset_index(drop=True)

    assert df.participant_id.is_unique
    data[name] = df

reference = data["RF"]

assert len(reference) == 54

y = reference["label"].to_numpy(dtype=int)

assert sum(y == 0) == 12
assert sum(y == 1) == 42

for name in names:
    df = data[name]

    assert df.participant_id.tolist() == (
        reference.participant_id.tolist()
    )

    assert np.array_equal(
        df["label"].to_numpy(dtype=int), y
    )

# 95% stratified bootstrap
rng = np.random.default_rng(42)
n_boot = 1000

healthy_idx = np.where(y == 0)[0]
pd_idx = np.where(y == 1)[0]

results = []
bootstrap_scores = {}

for name in names:

    df = data[name]
    probabilities = df[
        "probability_pd"
    ].to_numpy(dtype=float)

    threshold = float(
        df["locked_threshold"].iloc[0]
    )

    predictions = (
        probabilities >= threshold
    ).astype(int)

    def balanced_acc(indices):
        actual = y[indices]
        predicted = predictions[indices]

        sensitivity = np.mean(
            predicted[actual == 1] == 1
        )

        specificity = np.mean(
            predicted[actual == 0] == 0
        )

        return (sensitivity + specificity) / 2

    all_indices = np.arange(len(y))

    ba = balanced_acc(all_indices)
    auc = roc_auc_score(y, probabilities)

    ba_samples = []
    auc_samples = []

    # The same random seed produces reproducible CIs
    local_rng = np.random.default_rng(42)

    for _ in range(n_boot):

        sampled_hc = local_rng.choice(
            healthy_idx,
            size=len(healthy_idx),
            replace=True
        )

        sampled_pd = local_rng.choice(
            pd_idx,
            size=len(pd_idx),
            replace=True
        )

        sampled = np.concatenate([
            sampled_hc, sampled_pd
        ])

        ba_samples.append(
            balanced_acc(sampled)
        )

        auc_samples.append(
            roc_auc_score(
                y[sampled],
                probabilities[sampled]
            )
        )

    ba_ci = np.percentile(
        ba_samples, [2.5, 97.5]
    )

    auc_ci = np.percentile(
        auc_samples, [2.5, 97.5]
    )

    results.append({
        "Model": name,
        "Test_BA": ba,
        "BA_CI_Lower": ba_ci[0],
        "BA_CI_Upper": ba_ci[1],
        "Test_ROC_AUC": auc,
        "AUC_CI_Lower": auc_ci[0],
        "AUC_CI_Upper": auc_ci[1]
    })

    bootstrap_scores[name] = {
        "BA": np.array(ba_samples),
        "AUC": np.array(auc_samples)
    }

summary = pd.DataFrame(results)

summary.to_csv(
    OUT / "five_model_bootstrap_95CI.csv",
    index=False
)

print("FINAL ML RELIABILITY ANALYSIS")
print("Test participants:", len(y))
print("Bootstrap repetitions:", n_boot)

print(
    summary.round(3).to_string(index=False)
)

# Paired descriptive difference
difference = (
    bootstrap_scores["Hybrid CNN"]["BA"]
    - bootstrap_scores["RF"]["BA"]
)

print("\nHYBRID CNN vs RF")
print(
    "BA difference 95% interval:",
    np.round(
        np.percentile(difference, [2.5, 97.5]),
        3
    )
)

print("\nRESULTS SAVED SUCCESSFULLY!")
print("This is uncertainty estimation, not new validation.")
```

```python

import pandas as pd
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import roc_auc_score, balanced_accuracy_score
from xgboost import XGBClassifier
from pads_motor.evaluate import aggregate_by_participant
from scripts.train_classical import META_COLS

features = pd.read_csv(
    "outputs/features.csv", dtype={"participant_id": str}
)
splits = pd.read_csv(
    "outputs/participant_splits.csv",
    dtype={"participant_id": str}
)

data = features.merge(
    splits[["participant_id", "split"]],
    on="participant_id",
    validate="many_to_one"
)

train = data[data["split"] == "train"].reset_index(drop=True)
nums = [c for c in train.columns if c not in META_COLS]

ratio = (train.label == 0).sum() / (train.label == 1).sum()

models = {
    "RandomForest": RandomForestClassifier(
        n_estimators=200, min_samples_leaf=2,
        class_weight="balanced", random_state=42, n_jobs=-1
    ),
    "SVM": SVC(
        kernel="rbf", C=1.0, class_weight="balanced",
        probability=True, random_state=42
    ),
    "XGBoost": XGBClassifier(
        n_estimators=180, max_depth=3,
        learning_rate=0.05, scale_pos_weight=ratio,
        eval_metric="logloss", random_state=42, n_jobs=2
    )
}

cv = StratifiedGroupKFold(
    n_splits=3, shuffle=True, random_state=42
)

results = []

for name, classifier in models.items():
    for fold, (tr_idx, va_idx) in enumerate(
        cv.split(train, train.label, groups=train.participant_id),
        start=1
    ):
        tr = train.iloc[tr_idx]
        va = train.iloc[va_idx]

        pre = ColumnTransformer([
            ("numeric", StandardScaler(), nums),
            ("categorical", OneHotEncoder(
                handle_unknown="ignore"
            ), ["task", "wrist"])
        ])

        model = Pipeline([
            ("prepare", pre),
            ("classifier", classifier)
        ])

        model.fit(tr, tr.label)
        pd_index = list(
            model.named_steps["classifier"].classes_
        ).index(1)

        probs = model.predict_proba(va)[:, pd_index]
        subjects = aggregate_by_participant(va, probs)

        y_true = subjects["label"].astype(int)
        y_pred = (subjects["probability_pd"] >= 0.5).astype(int)

        results.append({
            "model": name,
            "fold": fold,
            "participants": len(subjects),
            "ROC_AUC": roc_auc_score(
                y_true, subjects["probability_pd"]
            ),
            "Balanced_Accuracy_0.5": balanced_accuracy_score(
                y_true, y_pred
            )
        })

cv_results = pd.DataFrame(results)

print("TRAINING-ONLY CROSS-VALIDATION")
print(cv_results.round(3).to_string(index=False))

print("\nAVERAGE BY MODEL")
print(
    cv_results.groupby("model")[
        ["ROC_AUC", "Balanced_Accuracy_0.5"]
    ].agg(["mean", "std"]).round(3).to_string()
)

cv_results.to_csv(
    "outputs/training_only_cv_results.csv", index=False
)
```

```python

import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    balanced_accuracy_score, recall_score, roc_auc_score
)
from pads_motor.evaluate import aggregate_by_participant

results = []

for name, classifier in models.items():
    fold_predictions = []

    for tr_idx, va_idx in cv.split(
        train, train.label, groups=train.participant_id
    ):
        tr = train.iloc[tr_idx]
        va = train.iloc[va_idx]

        pre = ColumnTransformer([
            ("numeric", StandardScaler(), nums),
            ("categorical",
             OneHotEncoder(handle_unknown="ignore"),
             ["task", "wrist"])
        ])

        clf = clone(classifier)

        if name == "XGBoost":
            n_healthy = (tr.label == 0).sum()
            n_pd = (tr.label == 1).sum()
            clf.set_params(
                scale_pos_weight=n_healthy / n_pd
            )

        pipeline = Pipeline([
            ("prepare", pre),
            ("classifier", clf)
        ])

        pipeline.fit(tr, tr.label)

        pd_idx = list(
            pipeline.named_steps["classifier"].classes_
        ).index(1)

        probabilities = pipeline.predict_proba(va)[:, pd_idx]
        subjects = aggregate_by_participant(
            va, probabilities
        )

        fold_predictions.append(subjects)

    oof = pd.concat(fold_predictions, ignore_index=True)
    y = oof["label"].astype(int)
    p = oof["probability_pd"].values

    best_threshold = 0.5
    best_ba = -1

    for threshold in np.arange(0.10, 0.96, 0.01):
        pred = (p >= threshold).astype(int)
        ba = balanced_accuracy_score(y, pred)

        if ba > best_ba:
            best_ba = ba
            best_threshold = threshold

    pred = (p >= best_threshold).astype(int)

    results.append({
        "Model": name,
        "CV_ROC_AUC": round(roc_auc_score(y, p), 3),
        "Threshold": round(best_threshold, 2),
        "Balanced_Accuracy": round(best_ba, 3),
        "PD_Sensitivity": round(
            recall_score(y, pred, pos_label=1), 3
        ),
        "Healthy_Specificity": round(
            recall_score(y, pred, pos_label=0), 3
        )
    })

summary = pd.DataFrame(results)

print("TRAINING-ONLY THRESHOLD EXPLORATION")
print(summary.to_string(index=False))

summary.to_csv(
    "outputs/cv_threshold_exploration.csv", index=False
)
```

```python

import numpy as np
import pandas as pd
from scipy.signal import welch
from pathlib import Path

rng = np.random.default_rng(42)
rows = []

rating_keys = [
    "updrs20_right", "updrs20_left",
    "updrs21_right", "updrs21_left"
]

for pid, record in tremor_data.items():
    if not isinstance(record[1], dict):
        continue

    labels = record[1]
    candidates = []

    for session_idx, session in enumerate(record[3]):
        arr = np.asarray(session)
        if arr.ndim == 3 and arr.shape[1:] == (500, 3):
            for k in range(len(arr) // 2):
                candidates.append((session_idx, k))

    if not candidates:
        continue

    chosen = rng.choice(
        len(candidates),
        size=min(20, len(candidates)),
        replace=False
    )

    measurements = []

    for idx in chosen:
        session_idx, k = candidates[int(idx)]
        session = np.asarray(record[3][session_idx])

        signal = np.concatenate(
            [session[2*k], session[2*k+1]], axis=0
        )

        if not np.isfinite(signal).all():
            continue

        magnitude = np.linalg.norm(signal, axis=1)
        frequencies, psd = welch(
            signal, fs=100, nperseg=500, axis=0
        )
        power = psd.sum(axis=1)

        tremor_band = power[
            (frequencies >= 3) & (frequencies <= 7)
        ].sum()
        reference_band = power[
            (frequencies >= 1) & (frequencies <= 15)
        ].sum()

        measurements.append({
            "rms": np.sqrt(np.mean(magnitude**2)),
            "std": np.std(magnitude),
            "tremor_band_ratio": (
                tremor_band / (reference_band + 1e-12)
            )
        })

    if not measurements:
        continue

    features = pd.DataFrame(measurements)
    severity = max(int(labels[k]) for k in rating_keys)

    rows.append({
        "participant_id": str(pid),
        "tremor_present": int(severity > 0),
        "max_clinical_rating": severity,
        "n_windows_used": len(measurements),
        "rms_median": features["rms"].median(),
        "rms_p90": features["rms"].quantile(0.9),
        "std_median": features["std"].median(),
        "band_ratio_median":
            features["tremor_band_ratio"].median(),
        "band_ratio_p90":
            features["tremor_band_ratio"].quantile(0.9)
    })

ts_features = pd.DataFrame(rows)

print("PARTICIPANT FEATURE TABLE")
print("Participants:", len(ts_features))
print("Missing values:", int(ts_features.isna().sum().sum()))
print("\nBinary label counts:")
print(ts_features["tremor_present"].value_counts())
print("\nFirst five rows:")
print(ts_features.head().to_string(index=False))

out = Path("/content/Zenodo_Tremor/ts_subject_features.csv")
ts_features.to_csv(out, index=False)
print("\nSaved:", out)
```

```python

import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    balanced_accuracy_score,
    recall_score,
    roc_auc_score,
    f1_score
)
from xgboost import XGBClassifier

# Load TS participant-level features
data = pd.read_csv(
    "/content/Zenodo_Tremor/ts_subject_features.csv"
)

feature_cols = [
    "rms_median",
    "rms_p90",
    "std_median",
    "band_ratio_median",
    "band_ratio_p90"
]

assert data["participant_id"].is_unique
assert data[feature_cols].notna().all().all()

X = data[feature_cols]
y = data["tremor_present"].astype(int)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []

for name in ["RandomForest", "SVM", "XGBoost"]:

    for fold, (train_idx, val_idx) in enumerate(
        cv.split(X, y), start=1
    ):

        X_train = X.iloc[train_idx]
        X_val = X.iloc[val_idx]

        y_train = y.iloc[train_idx]
        y_val = y.iloc[val_idx]

        if name == "RandomForest":
            model = RandomForestClassifier(
                n_estimators=200,
                min_samples_leaf=2,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )

        elif name == "SVM":
            model = make_pipeline(
                StandardScaler(),
                SVC(
                    kernel="rbf",
                    C=1.0,
                    class_weight="balanced",
                    probability=True,
                    random_state=42
                )
            )

        else:
            weight = (
                (y_train == 0).sum() /
                (y_train == 1).sum()
            )

            model = XGBClassifier(
                n_estimators=120,
                max_depth=2,
                learning_rate=0.05,
                scale_pos_weight=weight,
                eval_metric="logloss",
                random_state=42,
                n_jobs=2
            )

        model.fit(X_train, y_train)

        prob = model.predict_proba(X_val)[:, 1]
        pred = (prob >= 0.5).astype(int)

        results.append({
            "Model": name,
            "Fold": fold,
            "Balanced_Accuracy": balanced_accuracy_score(
                y_val, pred
            ),
            "Sensitivity": recall_score(
                y_val, pred, pos_label=1
            ),
            "Specificity": recall_score(
                y_val, pred, pos_label=0
            ),
            "ROC_AUC": roc_auc_score(y_val, prob),
            "Macro_F1": f1_score(
                y_val, pred, average="macro"
            )
        })

results_df = pd.DataFrame(results)

summary = results_df.groupby("Model")[
    [
        "Balanced_Accuracy",
        "Sensitivity",
        "Specificity",
        "ROC_AUC",
        "Macro_F1"
    ]
].mean().round(3)

print("TS TREMOR - 5-FOLD CV RESULTS")
print(summary.to_string())

results_df.to_csv(
    "/content/Zenodo_Tremor/ts_cv_fold_results.csv",
    index=False
)

summary.to_csv(
    "/content/Zenodo_Tremor/ts_cv_summary.csv"
)

print("\nTS cross-validation completed!")
```

```python

import numpy as np
import pandas as pd

from sklearn.model_selection import GroupKFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.base import clone
from sklearn.metrics import (
    mean_absolute_error,
    cohen_kappa_score,
    accuracy_score
)
from xgboost import XGBRegressor

folder = "/content/BS_Finger_Tapping"
bs = pd.read_csv(f"{folder}/bs_finger_tapping.csv")

# Same participant-ID convention as ROC-HCI
def get_subject_id(name):
    parts = str(name).split("-")
    return parts[0] if str(name).startswith("NIH") else parts[-3]

bs["subject_id"] = bs["filename"].apply(get_subject_id)

feature_cols = [
    "aperiodicity_denoised",
    "periodEntropy_denoised",
    "periodVarianceNorm_denoised",
    "period_median_denoised",
    "period_quartile_range_denoised",
    "period_min_denoised",
    "frequency_min_denoised",
    "frequency_stdev_denoised",
    "frequency_lr_slope_denoised",
    "numInterruptions_denoised",
    "num_peaks_trimmed"
]

X = bs[feature_cols].astype(float)
y = bs["Rating"].astype(int).to_numpy()
groups = bs["subject_id"].to_numpy()

assert X.notna().all().all()
assert bs["subject_id"].notna().all()
assert set(np.unique(y)).issubset({0, 1, 2, 3, 4})

print("Recordings:", len(bs))
print("Unique participants:", bs["subject_id"].nunique())
print("Score-4 participants:",
      bs.loc[bs.Rating == 4, "subject_id"].nunique())

models = {
    "RandomForest": RandomForestRegressor(
        n_estimators=200,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),
    "SVR": make_pipeline(
        StandardScaler(),
        SVR(kernel="rbf", C=1.0, epsilon=0.2)
    ),
    "XGBoost": XGBRegressor(
        n_estimators=200,
        max_depth=3,
        learning_rate=0.05,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=2
    )
}

cv = GroupKFold(n_splits=5)
fold_results = []
all_predictions = []

for name, estimator in models.items():
    for fold, (tr, va) in enumerate(
        cv.split(X, y, groups), start=1
    ):
        # Verify no participant appears in both sets
        assert set(groups[tr]).isdisjoint(set(groups[va]))

        model = clone(estimator)
        model.fit(X.iloc[tr], y[tr])

        continuous = np.clip(
            model.predict(X.iloc[va]), 0, 4
        )
        predicted = np.rint(continuous).astype(int)

        fold_results.append({
            "Model": name,
            "Fold": fold,
            "Participants": len(set(groups[va])),
            "MAE": mean_absolute_error(y[va], predicted),
            "Accuracy": accuracy_score(y[va], predicted),
            "QWK": cohen_kappa_score(
                y[va], predicted,
                labels=[0, 1, 2, 3, 4],
                weights="quadratic"
            )
        })

        all_predictions.extend([
            {
                "Model": name,
                "subject_id": groups[idx],
                "Actual": int(y[idx]),
                "Predicted": int(predicted[j])
            }
            for j, idx in enumerate(va)
        ])

fold_df = pd.DataFrame(fold_results)
pred_df = pd.DataFrame(all_predictions)

summary_rows = []

for name in models:
    p = pred_df[pred_df.Model == name]

    # Equal total weight per participant
    counts = p["subject_id"].value_counts()
    weights = 1 / p["subject_id"].map(counts)

    summary_rows.append({
        "Model": name,
        "MAE": mean_absolute_error(
            p.Actual, p.Predicted
        ),
        "Subject_Balanced_MAE": mean_absolute_error(
            p.Actual, p.Predicted,
            sample_weight=weights
        ),
        "Accuracy": accuracy_score(
            p.Actual, p.Predicted
        ),
        "Quadratic_Weighted_Kappa": cohen_kappa_score(
            p.Actual, p.Predicted,
            labels=[0, 1, 2, 3, 4],
            weights="quadratic"
        )
    })

summary = pd.DataFrame(summary_rows)

print("\nBS SEVERITY — GROUPED 5-FOLD CV RESULTS")
print(summary.round(3).to_string(index=False))

fold_df.to_csv(
    f"{folder}/bs_cv_folds.csv", index=False
)
pred_df.to_csv(
    f"{folder}/bs_cv_predictions.csv", index=False
)
summary.to_csv(
    f"{folder}/bs_cv_summary.csv", index=False
)

print("\nBS training experiment completed!")
```

```python

import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import (
    balanced_accuracy_score,
    roc_auc_score,
    recall_score,
    f1_score
)
from xgboost import XGBClassifier

# Previously restored PADS project
root = Path("/content/PADS_RF_SVM_XGBoost_Starter")

features_path = root / "outputs/features.csv"
splits_path = root / "outputs/participant_splits.csv"

assert features_path.exists(), "Restore PADS backup first."
assert splits_path.exists(), "Missing participant splits."

features = pd.read_csv(
    features_path, dtype={"participant_id": str}
)
splits = pd.read_csv(
    splits_path, dtype={"participant_id": str}
)

data = features.merge(
    splits[["participant_id", "split"]],
    on="participant_id",
    validate="many_to_one"
)

# Training subjects ONLY: do not reuse old test set
train = data[data["split"] == "train"].copy()

# Gyroscope-only features
rs_features = [
    "gyro_mean_rad_s",
    "gyro_rms_rad_s",
    "gyro_std_rad_s",
    "gyro_p95_rad_s",
    "gyro_peak_rad_s",
    "gyro_peak_to_peak_rad_s",
    "gyro_x_std_rad_s",
    "gyro_y_std_rad_s",
    "gyro_z_std_rad_s"
]

assert all(c in train.columns for c in rs_features)
assert train[rs_features].notna().all().all()

X = train[rs_features].astype(float)
y = train["label"].astype(int)
groups = train["participant_id"]

print("RS TRAINING DATA")
print("Recordings:", len(train))
print("Participants:", groups.nunique())
print("Gyroscope features:", len(rs_features))

print("\nPD / Healthy participants:")
print(
    train.groupby("label")["participant_id"].nunique()
)

models = {
    "RandomForest": RandomForestClassifier(
        n_estimators=200,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),
    "SVM": make_pipeline(
        StandardScaler(),
        SVC(
            kernel="rbf",
            C=1.0,
            class_weight="balanced",
            probability=True,
            random_state=42
        )
    ),
    "XGBoost": XGBClassifier(
        n_estimators=180,
        max_depth=3,
        learning_rate=0.05,
        eval_metric="logloss",
        random_state=42,
        n_jobs=2
    )
}

cv = StratifiedGroupKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

results = []

for name, base_model in models.items():

    for fold, (tr, va) in enumerate(
        cv.split(X, y, groups), start=1
    ):

        assert set(groups.iloc[tr]).isdisjoint(
            set(groups.iloc[va])
        )

        if name == "RandomForest":
            model = RandomForestClassifier(
                n_estimators=200,
                min_samples_leaf=2,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )

        elif name == "SVM":
            model = make_pipeline(
                StandardScaler(),
                SVC(
                    kernel="rbf",
                    C=1.0,
                    class_weight="balanced",
                    probability=True,
                    random_state=42
                )
            )

        else:
            healthy = (y.iloc[tr] == 0).sum()
            pd_count = (y.iloc[tr] == 1).sum()

            model = XGBClassifier(
                n_estimators=180,
                max_depth=3,
                learning_rate=0.05,
                scale_pos_weight=healthy / pd_count,
                eval_metric="logloss",
                random_state=42,
                n_jobs=2
            )

        model.fit(X.iloc[tr], y.iloc[tr])

        pd_idx = list(model.classes_).index(1)
        probs = model.predict_proba(
            X.iloc[va]
        )[:, pd_idx]

        # One prediction per participant
        subject_predictions = pd.DataFrame({
            "participant_id": groups.iloc[va].values,
            "label": y.iloc[va].values,
            "probability_pd": probs
        }).groupby(
            "participant_id", as_index=False
        ).agg({
            "label": "first",
            "probability_pd": "mean"
        })

        actual = subject_predictions["label"]
        probability = subject_predictions["probability_pd"]
        predicted = (probability >= 0.5).astype(int)

        results.append({
            "Model": name,
            "Fold": fold,
            "Participants": len(subject_predictions),
            "Balanced_Accuracy":
                balanced_accuracy_score(actual, predicted),
            "Sensitivity":
                recall_score(
                    actual, predicted, pos_label=1,
                    zero_division=0
                ),
            "Specificity":
                recall_score(
                    actual, predicted, pos_label=0,
                    zero_division=0
                ),
            "ROC_AUC":
                roc_auc_score(actual, probability),
            "Macro_F1":
                f1_score(
                    actual, predicted,
                    average="macro",
                    zero_division=0
                )
        })

rs_results = pd.DataFrame(results)

rs_summary = rs_results.groupby("Model")[
    [
        "Balanced_Accuracy",
        "Sensitivity",
        "Specificity",
        "ROC_AUC",
        "Macro_F1"
    ]
].mean().round(3)

print("\nRS GYROSCOPE — 3-FOLD CV RESULTS")
print(rs_summary.to_string())

# Save preliminary evaluation results
out = Path("/content/RS_Gyroscope")
out.mkdir(exist_ok=True)

rs_results.to_csv(
    out / "rs_cv_fold_results.csv",
    index=False
)

rs_summary.to_csv(
    out / "rs_cv_summary.csv"
)

print("\nRS training experiment completed!")
```

```python

import re
import zipfile
import numpy as np
import pandas as pd

from pathlib import Path
from scipy.signal import welch
from scipy.ndimage import uniform_filter1d

root = Path("/content/GI_MyGait")
zip_path = root / "Parkinson_Dataset.zip"

rows = []
skipped = []

# Indoor tests only: Healthy and Parkinson's
indoor_tests = {
    "9_6min_in": 0,
    "20_6min_in": 1
}

with zipfile.ZipFile(zip_path) as z:

    files_to_process = []

    for name in z.namelist():
        parts = name.split("/")

        if (
            len(parts) >= 4
            and parts[2] in indoor_tests
            and name.lower().endswith(".csv")
        ):
            files_to_process.append(name)

    print("Indoor recordings:", len(files_to_process))

    for index, name in enumerate(
        sorted(files_to_process), start=1
    ):
        parts = name.split("/")
        label = indoor_tests[parts[2]]

        # Handles 01RTSOCS, 01-RTSOCS,
        # and 01-1RTSOCS filename formats.
        filename = Path(name).name
        match = re.match(
            r"^(\d{2})(?:-\d+)?-?RTSOCS",
            filename
        )

        if match is None:
            skipped.append((name, "Unknown participant"))
            continue

        participant_id = (
            f"{'PD' if label else 'HC'}_"
            f"{match.group(1)}"
        )

        with z.open(name) as f:
            df = pd.read_csv(
                f,
                usecols=[
                    "foot", "timestamp",
                    "acx", "acy", "acz"
                ]
            )

        # Use the same foot for everyone
        df = df[
            df["foot"].astype(str).str.lower() == "left"
        ].copy()

        for c in ["timestamp", "acx", "acy", "acz"]:
            df[c] = pd.to_numeric(
                df[c], errors="coerce"
            )

        df = df.dropna().sort_values("timestamp")

        if df.empty:
            skipped.append((name, "No valid samples"))
            continue

        start = df["timestamp"].iloc[0]

        # Exactly the device's 180-second window
        df = df[
            (df["timestamp"] >= start)
            & (df["timestamp"] < start + 180)
        ]

        t = df["timestamp"].to_numpy()
        acc = df[["acx", "acy", "acz"]].to_numpy()

        if len(t) < 4000 or t[-1] - t[0] < 179.5:
            skipped.append((name, "Short recording"))
            continue

        dt = np.diff(t)
        dt = dt[dt > 0]

        if len(dt) == 0 or np.max(dt) > 1.0:
            skipped.append((name, "Timestamp gap"))
            continue

        fs = 1.0 / np.median(dt)

        if not 20 <= fs <= 200:
            skipped.append((name, "Sampling rate"))
            continue

        # Rotation-invariant acceleration magnitude
        magnitude = np.linalg.norm(acc, axis=1)

        # Remove slow baseline movement
        baseline = uniform_filter1d(
            magnitude,
            size=max(3, int(fs)),
            mode="nearest"
        )

        dynamic = magnitude - baseline

        # Gait-related spectral analysis
        frequencies, power = welch(
            dynamic,
            fs=fs,
            nperseg=min(len(dynamic), int(fs * 8))
        )

        gait_mask = (
            (frequencies >= 0.5)
            & (frequencies <= 3.0)
        )

        reference_mask = (
            (frequencies >= 0.3)
            & (frequencies <= 10.0)
        )

        gait_power = np.trapezoid(
            power[gait_mask],
            frequencies[gait_mask]
        )

        reference_power = np.trapezoid(
            power[reference_mask],
            frequencies[reference_mask]
        )

        dominant_frequency = frequencies[
            gait_mask
        ][np.argmax(power[gait_mask])]

        rows.append({
            "participant_id": participant_id,
            "label": label,
            "acc_dynamic_rms": float(
                np.sqrt(np.mean(dynamic ** 2))
            ),
            "acc_dynamic_std": float(
                np.std(dynamic)
            ),
            "acc_dynamic_p90": float(
                np.percentile(np.abs(dynamic), 90)
            ),
            "acc_dynamic_iqr": float(
                np.percentile(dynamic, 75)
                - np.percentile(dynamic, 25)
            ),
            "dominant_gait_frequency_hz": float(
                dominant_frequency
            ),
            "gait_band_power_ratio": float(
                gait_power / (reference_power + 1e-12)
            )
        })

        if index % 10 == 0:
            print(
                f"Processed {index}/"
                f"{len(files_to_process)}"
            )

gi_features = pd.DataFrame(rows)

if not gi_features.empty:
    assert gi_features["participant_id"].is_unique
    assert gi_features.isna().sum().sum() == 0

    print("\nGI FEATURE EXTRACTION RESULTS")
    print("Participants:", len(gi_features))

    print("\nHealthy / Parkinson's counts:")
    print(
        gi_features["label"].value_counts()
        .sort_index()
    )

    print("\nFeature columns:")
    print(gi_features.columns.tolist())

    print("\nFirst five rows:")
    print(gi_features.head().to_string(index=False))

    output = root / "gi_180s_features.csv"
    gi_features.to_csv(output, index=False)

    print("\nSaved:", output)

print("\nSkipped recordings:", len(skipped))
for item in skipped[:10]:
    print(item)

print("\nGI preprocessing step finished!")
```

```python

import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    balanced_accuracy_score, roc_auc_score,
    recall_score, f1_score
)
from xgboost import XGBClassifier

# Load uploaded wrist-gait features
df = pd.read_csv(
    "/content/GI_Wrist_Features_51_Participants.csv"
)

features = [
    "acc_dynamic_rms_ms2",
    "acc_dynamic_p90_ms2",
    "acc_dynamic_jerk_rms_ms3",
    "gait_dominant_freq_hz",
    "gait_bandpower_ratio"
]

X = df[features].astype(float)
y = df["label"].astype(int)

assert df["participant_id"].is_unique
assert set(y.unique()) == {0, 1}
assert np.isfinite(X.to_numpy()).all()

print("GI WRIST DATASET")
print("Participants:", len(df))
print("Healthy:", (y == 0).sum())
print("Parkinson's:", (y == 1).sum())

models = {
    "RandomForest": RandomForestClassifier(
        n_estimators=200,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),
    "SVM": make_pipeline(
        StandardScaler(),
        SVC(
            kernel="rbf",
            C=1.0,
            class_weight="balanced",
            probability=True,
            random_state=42
        )
    ),
    "XGBoost": XGBClassifier(
        n_estimators=180,
        max_depth=2,
        learning_rate=0.05,
        eval_metric="logloss",
        random_state=42,
        n_jobs=2
    )
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []

for name, base_model in models.items():

    for fold, (tr, va) in enumerate(
        cv.split(X, y), start=1
    ):
        model = clone(base_model)

        if name == "XGBoost":
            weight = (
                (y.iloc[tr] == 0).sum()
                / (y.iloc[tr] == 1).sum()
            )
            model.set_params(
                scale_pos_weight=weight
            )

        model.fit(X.iloc[tr], y.iloc[tr])

        probabilities = model.predict_proba(
            X.iloc[va]
        )[:, 1]

        predicted = (
            probabilities >= 0.5
        ).astype(int)

        results.append({
            "Model": name,
            "Fold": fold,
            "Balanced_Accuracy":
                balanced_accuracy_score(
                    y.iloc[va], predicted
                ),
            "PD_Sensitivity":
                recall_score(
                    y.iloc[va], predicted,
                    pos_label=1
                ),
            "Healthy_Specificity":
                recall_score(
                    y.iloc[va], predicted,
                    pos_label=0
                ),
            "ROC_AUC":
                roc_auc_score(
                    y.iloc[va], probabilities
                ),
            "Macro_F1":
                f1_score(
                    y.iloc[va], predicted,
                    average="macro"
                )
        })

    print(name, "completed!")

results = pd.DataFrame(results)

summary = (
    results.groupby("Model")
    .mean(numeric_only=True)
    .drop(columns="Fold")
    .round(3)
)

print("\nGI WRIST - 5-FOLD CV RESULTS")
print(summary.to_string())

folder = Path("/content/GI_Wrist")
folder.mkdir(exist_ok=True)

results.to_csv(
    folder / "gi_wrist_cv_folds.csv",
    index=False
)

summary.to_csv(
    folder / "gi_wrist_cv_summary.csv"
)

print("\nGI wrist training experiment completed!")
```

```python

import pandas as pd
import numpy as np
import joblib
import json
import zipfile

from pathlib import Path
from google.colab import files
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from xgboost import XGBClassifier

folder = Path("/content/GI_Wrist")
save_dir = folder / "saved_models"
save_dir.mkdir(parents=True, exist_ok=True)

data = pd.read_csv(
    "/content/GI_Wrist_Features_51_Participants.csv"
)

feature_cols = [
    "acc_dynamic_rms_ms2",
    "acc_dynamic_p90_ms2",
    "acc_dynamic_jerk_rms_ms3",
    "gait_dominant_freq_hz",
    "gait_bandpower_ratio"
]

X = data[feature_cols].astype(float)
y = data["label"].astype(int)

assert data["participant_id"].is_unique
assert len(data) == 51
assert np.isfinite(X.to_numpy()).all()

weight = (y == 0).sum() / (y == 1).sum()

models = {
    "RandomForest": RandomForestClassifier(
        n_estimators=200,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ),
    "SVM": make_pipeline(
        StandardScaler(),
        SVC(
            kernel="rbf",
            C=1.0,
            class_weight="balanced",
            probability=True,
            random_state=42
        )
    ),
    "XGBoost": XGBClassifier(
        n_estimators=180,
        max_depth=2,
        learning_rate=0.05,
        scale_pos_weight=weight,
        eval_metric="logloss",
        random_state=42,
        n_jobs=2
    )
}

# Refit each model on all 51 subjects
# after cross-validation evaluation
for name, model in models.items():
    model.fit(X, y)
    path = save_dir / f"GI_{name}.joblib"
    joblib.dump(model, path)
    print("Saved:", path.name)

metadata = {
    "dataset": "WearGait-PD",
    "task": "Preliminary PD vs Healthy classification",
    "participants": 51,
    "features": feature_cols,
    "sensor": "Wrist accelerometer",
    "protocol": "Short walking trials",
    "clinical_gait_0_to_4_validated": False,
    "device_180s_validated": False
}

(save_dir / "metadata.json").write_text(
    json.dumps(metadata, indent=2)
)

backup = Path("/content/GI_Wrist_Models_Backup.zip")

with zipfile.ZipFile(
    backup, "w", zipfile.ZIP_DEFLATED
) as z:

    for f in save_dir.iterdir():
        z.write(f, f"saved_models/{f.name}")

    for name in [
        "gi_wrist_cv_folds.csv",
        "gi_wrist_cv_summary.csv"
    ]:
        f = folder / name
        if f.exists():
            z.write(f, name)

print("\nGI BACKUP CREATED SUCCESSFULLY!")
files.download(str(backup))
```
